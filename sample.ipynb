```python
import glob
import os
import pandas as pd

# 1. Set your folder path here (e.g., './my_data_folder' or absolute path)
folder_path = "./"

# 2. Get all CSV files in the folder
csv_files = glob.glob(os.path.join(folder_path, "*.csv"))

sampled_dfs = []

for file in csv_files:
  try:
    # Read the CSV file
    df = pd.read_csv(file)

    # Optional: Add a column to track the original source file name
    df["source_file"] = os.path.basename(file)

    # 3. Take a sample from each file
    # Change n=5 to the number of rows you want per file, or use frac=0.1 for 10%
    sample_size = min(5, len(df))
    # df_sample = df.sample(n=sample_size, random_state=42)
    df_sample = df.sample(frac=0.1, random_state=42)

    sampled_dfs.append(df_sample)
  except Exception as e:
    print(f"Error reading {file}: {e}")

# 4. Combine all samples and save to sample.csv
if sampled_dfs:
  final_sample = pd.concat(sampled_dfs, ignore_index=True)
  final_sample.to_csv("sample.csv", index=False)
  print(
      "Successfully combined samples from"
      f" {len(sampled_dfs)} file(s) and saved to 'sample.csv'."
  )
else:
  print("No CSV files were found or successfully processed.")
```
